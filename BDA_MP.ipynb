```python
%%writefile /content/project/parse.pig
raw = LOAD '/content/project/ssh.log' USING TextLoader() AS (line:chararray);

attempts = FILTER raw BY line MATCHES '.*(Failed password|Invalid user).* from [0-9.]+.*';

parsed = FOREACH attempts GENERATE
    REGEX_EXTRACT(line, '^(\\w+ +\\d+) ', 1) AS day,
    (int)REGEX_EXTRACT(line, '^\\w+ +\\d+ (\\d+):', 1) AS hour,
    REGEX_EXTRACT(line, '(Failed password|Invalid user)', 1) AS event,
    REGEX_EXTRACT(line, '(?:for invalid user |for |Invalid user )(\\S+) from', 1) AS username,
    REGEX_EXTRACT(line, 'from ([0-9]{1,3}(?:\\.[0-9]{1,3}){3})', 1) AS ip;

STORE parsed INTO '/content/project/parsed' USING PigStorage(',');
```

```python
%%writefile /content/project/queries.hql
SET mapreduce.framework.name=local;
SET hive.exec.mode.local.auto=true;

DROP TABLE IF EXISTS ssh_logs;
CREATE EXTERNAL TABLE ssh_logs (
  log_date STRING, hr INT, event STRING, username STRING, ip STRING
)
ROW FORMAT DELIMITED FIELDS TERMINATED BY ','
LOCATION 'file:///content/project/parsed';

SELECT COUNT(*) AS total_rows FROM ssh_logs;

-- 1. Top attacking IPs
SELECT ip, COUNT(*) AS failures FROM ssh_logs
WHERE event = 'Failed password'
GROUP BY ip ORDER BY failures DESC LIMIT 10;

-- 2. Failed logins per hour
SELECT hr, COUNT(*) AS failures FROM ssh_logs
WHERE event = 'Failed password'
GROUP BY hr ORDER BY hr;

-- 3. Most targeted usernames
SELECT username, COUNT(*) AS attempts FROM ssh_logs
GROUP BY username ORDER BY attempts DESC LIMIT 10;

-- 4. Suspicious IP flag (more than 5 failures)
SELECT ip, COUNT(*) AS failures,
       CASE WHEN COUNT(*) > 5 THEN 'SUSPICIOUS' ELSE 'normal' END AS flag
FROM ssh_logs WHERE event = 'Failed password'
GROUP BY ip ORDER BY failures DESC;
```

```python
import subprocess, pandas as pd, io

def hive_query(sql):
    result = subprocess.run(
        ["bash", "-c", f'JAVA_HOME=/usr/lib/jvm/java-21-openjdk-amd64 '
         f'/content/apache-hive-4.2.1-bin/bin/hive '
         f'--hiveconf hive.execution.engine=mr '
         f'--hiveconf mapreduce.framework.name=local '
         f'--hiveconf hive.exec.mode.local.auto=true '
         f'-e "{sql}"'],
        capture_output=True, text=True)
    lines = [l for l in result.stdout.split("\n")
             if l.strip() and not any(x in l for x in ["WARN","INFO","SLF4J","OK","Time taken"])]
    return "\n".join(lines)

# Setup table (run once)
setup = """
DROP TABLE IF EXISTS ssh_logs;
CREATE TABLE ssh_logs (day STRING, hour INT, event STRING, username STRING, ip STRING)
ROW FORMAT DELIMITED FIELDS TERMINATED BY ',' STORED AS TEXTFILE;
LOAD DATA LOCAL INPATH '/content/project/parsed/part-m-00000' INTO TABLE ssh_logs;
"""
print(hive_query(setup))
```

```python
q1 = hive_query("SELECT ip, COUNT(*) as attempts FROM ssh_logs GROUP BY ip ORDER BY attempts DESC LIMIT 10")
q2 = hive_query("SELECT hour, COUNT(*) as attempts FROM ssh_logs GROUP BY hour ORDER BY hour")
q3 = hive_query("SELECT username, COUNT(*) as attempts FROM ssh_logs GROUP BY username ORDER BY attempts DESC LIMIT 10")
q4 = hive_query("SELECT ip, COUNT(*) as attempts FROM ssh_logs GROUP BY ip HAVING COUNT(*) > 10 ORDER BY attempts DESC")

def to_df(raw, cols):
    rows = [r.split("\t") for r in raw.strip().split("\n") if "\t" in r]
    return pd.DataFrame(rows, columns=cols)

top_ips    = to_df(q1, ["IP","Attempts"])
by_hour    = to_df(q2, ["Hour","Attempts"])
top_users  = to_df(q3, ["Username","Attempts"])
suspicious = to_df(q4, ["IP","Attempts"])

print("Top IPs:\n", top_ips)
print("\nBy Hour:\n", by_hour)
print("\nTop Usernames:\n", top_users)
print("\nSuspicious IPs:\n", suspicious.head())
```

```python
!pip install -q gradio

import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use("Agg")

# --- data already in memory from earlier ---
top_ips["Attempts"] = top_ips["Attempts"].astype(int)
by_hour["Attempts"] = by_hour["Attempts"].astype(int)
by_hour["Hour"] = by_hour["Hour"].astype(int)
top_users["Attempts"] = top_users["Attempts"].astype(int)
suspicious["Attempts"] = suspicious["Attempts"].astype(int)

def plot_top_ips():
    fig, ax = plt.subplots(figsize=(8,4))
    ax.barh(top_ips["IP"], top_ips["Attempts"], color="crimson")
    ax.set_xlabel("Attempts")
    ax.set_title("Top 10 Attacking IPs")
    ax.invert_yaxis()
    plt.tight_layout()
    return fig

def plot_by_hour():
    fig, ax = plt.subplots(figsize=(8,4))
    ax.bar(by_hour["Hour"].astype(str), by_hour["Attempts"], color="steelblue")
    ax.set_xlabel("Hour of Day")
    ax.set_ylabel("Attempts")
    ax.set_title("Attack Attempts by Hour")
    plt.tight_layout()
    return fig

def plot_top_users():
    fig, ax = plt.subplots(figsize=(8,4))
    ax.barh(top_users["Username"], top_users["Attempts"], color="darkorange")
    ax.set_xlabel("Attempts")
    ax.set_title("Top 10 Targeted Usernames")
    ax.invert_yaxis()
    plt.tight_layout()
    return fig

def ip_lookup(ip):
    ip = ip.strip()
    if not ip:
        return "Enter an IP address."
    count = df[df["ip"] == ip].shape[0]
    if count == 0:
        return f"IP {ip} — No attempts found."
    flag = "—— SUSPICIOUS" if count > 10 else "Low Activity"
    events = df[df["ip"] == ip]["event"].value_counts().to_dict()
    users  = df[df["ip"] == ip]["username"].value_counts().head(3).to_dict()
    return (f"IP: {ip}\nTotal Attempts: {count}\nStatus: {flag}\n"
            f"Events: {events}\nTop Targeted Users: {users}")

with gr.Blocks(title="SSH Intrusion Analytics Dashboard") as demo:
    gr.Markdown("# 🔒 SSH Intrusion Analytics Dashboard")
    gr.Markdown("**Pipeline: Flume → Pig → Hive → Dashboard** | Dataset: OpenSSH logs (633 records)")

    with gr.Tabs():
        with gr.Tab("Top Attacking IPs"):
            gr.Plot(plot_top_ips)
            gr.Dataframe(value=top_ips, label="Top 10 Attacking IPs")

        with gr.Tab("Attacks by Hour"):
            gr.Plot(plot_by_hour)
            gr.Dataframe(value=by_hour, label="Hourly Attack Distribution")

        with gr.Tab("Targeted Usernames"):
            gr.Plot(plot_top_users)
            gr.Dataframe(value=top_users, label="Top 10 Targeted Usernames")

        with gr.Tab("Suspicious IPs"):
            gr.Dataframe(value=suspicious, label="IPs with more than 10 attempts")

        with gr.Tab("IP Lookup"):
            gr.Markdown("Enter any IP to check its activity.")
            ip_input = gr.Textbox(label="IP Address", placeholder="e.g. 183.62.140.253")
            result   = gr.Textbox(label="Result", lines=6)
            btn      = gr.Button("Search")
            btn.click(fn=ip_lookup, inputs=ip_input, outputs=result)

demo.launch(share=True, debug=False, quiet=True)
```
